```python
import ipywidgets as widgets
from IPython.display import display
from pathlib import Path

candidate_yaml_path = Path("candidate.yaml")
current_content = candidate_yaml_path.read_text()

# Create a Textarea widget for editing
edit_area = widgets.Textarea(
    value=current_content,
    description='Edit candidate.yaml:',
    disabled=False,
    continuous_update=False,
    layout={'height': '300px', 'width': 'auto'}
)
display(edit_area)
```

```python
if 'edit_area' in locals():
    # Save the edited content back to the file
    candidate_yaml_path.write_text(edit_area.value)
    print("candidate.yaml updated successfully.")
else:
    print("Please run the previous cell first to initialize the editor.")
```

```python
import yaml
from pathlib import Path

candidate_yaml_path = Path("candidate.yaml")

# Load the existing candidate.yaml content
with open(candidate_yaml_path, 'r') as f:
    candidate_data = yaml.safe_load(f)

# Update the specified sections
candidate_data['target_roles'] = [
  "Technical Program Manager",
  "SAP Technical Program Manager",
  "AI Program Manager",
  "Program Manager",
  "Delivery Lead",
  "Senior Program Manager",
  "Technical Project Manager"
]

candidate_data['required_keywords'] = [
  "Program Manager",
  "TPM",
  "SAP"
]

candidate_data['preferred_keywords'] = [
  "S/4HANA",
  "BTP",
  "OData",
  "Agile",
  "Scrum",
  "Stakeholder Management",
  "Cross-functional",
  "AI/ML",
  "Delivery Management",
  "SDLC",
  "Vendor Management",
  "Risk Management",
  "PRINCE2",
  "CSM"
]

# Write the updated content back to candidate.yaml
with open(candidate_yaml_path, 'w') as f:
    yaml.dump(candidate_data, f, sort_keys=False)

print("candidate.yaml updated successfully with the new role and keyword preferences.")

# Display the updated content to confirm
print("\n--- Updated candidate.yaml content ---")
print(candidate_yaml_path.read_text())
```
